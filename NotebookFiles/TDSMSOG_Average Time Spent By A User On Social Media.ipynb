```python
import sys
from pathlib import Path

BASE_DIR = Path().resolve()             
CODES_DIR = BASE_DIR.parent / "Codes"
sys.path.insert(0, str(CODES_DIR))

import pandas as pd
import numpy as np
import dataset_read
import experiment
dataread = dataset_read.DATAREAD()
```

```python
csv = pd.read_csv(r"..\Datasets\dummy_data.csv")

dataread.label_encoding(csv,"gender")
dataread.label_encoding(csv,"platform")
dataread.label_encoding(csv,"interests")
dataread.label_encoding(csv,"location")
dataread.label_encoding(csv,"demographics")
dataread.label_encoding(csv,"income")
dataread.label_encoding(csv,"profession")
dataread.label_encoding(csv,"indebt")
dataread.label_encoding(csv,"isHomeOwner")
dataread.label_encoding(csv,"Owns_Car")

grouped = csv.groupby('time_spent')

for category, group in grouped:
    print(f"Category: {category}")
    print(len(group))
    
# Calculate sparsity
total_elements = csv.size
zero_elements = (csv == 0).sum().sum()
sparsity = (zero_elements / total_elements) * 100

print("Sparsity (%):", sparsity)

dataread.initializedataset(csv,"time_spent")
```
