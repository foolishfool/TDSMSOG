```python
import sys
from pathlib import Path

BASE_DIR = Path().resolve()             
CODES_DIR = BASE_DIR.parent / "Codes"
sys.path.insert(0, str(CODES_DIR))

import pandas as pd
import numpy as np
import dataset_read
import experiment
dataread = dataset_read.DATAREAD()
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
car_evaluation = fetch_ucirepo(id=19) 
  
# data (as pandas dataframes) 
X = car_evaluation.data.features 
y = car_evaluation.data.targets 

dataread.label_encoding(X,"buying")
dataread.label_encoding(X,"maint")
dataread.label_encoding(X,"doors")
dataread.label_encoding(X,"persons")
dataread.label_encoding(X,"lug_boot")
dataread.label_encoding(X,"safety")
dataread.label_encoding(y,"class")
grouped = y.groupby('class')

# Iterate through groups
for category, group in grouped:
    print(f"Category: {category}")
    print(len(group))
    
# Calculate sparsity
total_elements = X.size
zero_elements = (X == 0).sum().sum()
sparsity = (zero_elements / total_elements) * 100

#print("Class Distribution:", class_distribution)
print("Sparsity (%):", sparsity)
    
dataread.readucidata(X,y,"class")
```

```python

import experiment
class_num= 1
dim_num = X.shape[1]
scope_num = 2
experiment = experiment.Experiment()
experiment.Ttest(dataread,class_num, dim_num,scope_num,20,"Car")
```
