```python
import sys
from pathlib import Path

BASE_DIR = Path().resolve()             
CODES_DIR = BASE_DIR.parent / "Codes"
sys.path.insert(0, str(CODES_DIR))

import pandas as pd
import numpy as np
import dataset_read
import experiment
dataread = dataset_read.DATAREAD()
```

```python
csv = pd.read_csv(r"..\Datasets\Hotel Reservations.csv")
csv = csv.drop(columns=['Booking_ID'])
csv = csv.sample(int(csv.shape[0]))


dataread.label_encoding(csv,"type_of_meal_plan")
dataread.label_encoding(csv,"room_type_reserved")
dataread.label_encoding(csv,"arrival_year")
dataread.label_encoding(csv,"market_segment_type")
dataread.label_encoding(csv,"booking_status")


grouped = csv.groupby('booking_status')

# Iterate through groups
for category, group in grouped:
    print(f"Category: {category}")
    print(len(group))
    


# Calculate sparsity
total_elements = csv.size
zero_elements = (csv == 0).sum().sum()
sparsity = (zero_elements / total_elements) * 100

#print("Class Distribution:", class_distribution)
print("Sparsity (%):", sparsity)


dataread.initializedataset(csv,"booking_status")

```

```python
import experiment
class_num= 1
dim_num = 17
scope_num = 2
experiment = experiment.Experiment()
experiment.Ttest(dataread,class_num, dim_num,scope_num,20,"Hotel Reservations")
```
