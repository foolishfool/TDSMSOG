```python
import sys
from pathlib import Path

BASE_DIR = Path().resolve()             
CODES_DIR = BASE_DIR.parent / "Codes"
sys.path.insert(0, str(CODES_DIR))

import pandas as pd
import numpy as np
import dataset_read
import experiment
dataread = dataset_read.DATAREAD()
```

```python

csv = pd.read_csv(r"..\Datasets\bi.csv")

dataread.label_encoding(csv,"fNAME")
dataread.label_encoding(csv,"lNAME")
dataread.label_encoding(csv,"gender")
dataread.label_encoding(csv,"country")
dataread.label_encoding(csv,"residence")
dataread.label_encoding(csv,"prevEducation")

grouped = csv.groupby('prevEducation')

# Iterate through groups
for category, group in grouped:
    print(f"Category: {category}")
    print(len(group))
    


# Calculate sparsity
total_elements = csv.size
zero_elements = (csv == 0).sum().sum()
sparsity = (zero_elements / total_elements) * 100

#print("Class Distribution:", class_distribution)
print("Sparsity (%):", sparsity)


dataread.initializedataset(csv,"prevEducation")
```

```python
#confidence interval
import experiment
class_num= 1
dim_num = 10
scope_num = 2
experiment = experiment.Experiment()
experiment.Ttest(dataread,class_num, dim_num,scope_num,40,"BI")
```
