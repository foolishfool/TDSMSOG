```python
#THIS CODE ONLY DO THE WHOLE DATA TRAINING, NO CROSS_VALIDATION AND FEATURE SELECTION
import pandas as pd
import numpy as np
import dataset_read
import warnings
import experiment
warnings.filterwarnings("ignore", category=np.VisibleDeprecationWarning) 
dataread = dataset_read.DATAREAD()
```

```python
csv = pd.read_csv("Predict Hair Fall.csv")
csv = csv.drop(columns=['Id'])
#csv = csv.sample(int(csv.shape[0]*0.1))


dataread.label_encoding(csv,"Genetics")
dataread.label_encoding(csv,"Hormonal Changes")
dataread.label_encoding(csv,"Medical Conditions")
dataread.label_encoding(csv,"Medications & Treatments")
dataread.label_encoding(csv,"Nutritional Deficiencies ")
dataread.label_encoding(csv,"Stress")
dataread.label_encoding(csv,"Poor Hair Care Habits ")
dataread.label_encoding(csv,"Environmental Factors")
dataread.label_encoding(csv,"Smoking")
dataread.label_encoding(csv,"Weight Loss ")

grouped = csv.groupby('Hair Loss')

# Iterate through groups
for category, group in grouped:
    print(f"Category: {category}")
    print(len(group))
    


# Calculate sparsity
total_elements = csv.size
zero_elements = (csv == 0).sum().sum()
sparsity = (zero_elements / total_elements) * 100

#print("Class Distribution:", class_distribution)
print("Sparsity (%):", sparsity)
#csv_test = csv.sample(int(csv.shape[0]*0.3))
#csv_training = csv.drop(csv_test.index,axis = 0)

dataread.initializedataset(csv,"Hair Loss")

```

```python
import experiment
class_num= 1
dim_num = 11
scope_num = 2
experiment = experiment.Experiment()
experiment.Ttest(dataread,class_num, dim_num,scope_num,25,"HairFall")
```
